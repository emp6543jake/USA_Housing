### Step 1: Import Libraries
Imports essential libraries for data manipulation `pandas`, `numpy` `matplotlib` `seaborn` `sklearn`). It also imports `pymongo` to connect Python to MongoDB database.

```python
import pandas as pd
import numpy as np
import pymongo
```

### Step 2: Database Connection & Data Ingestion
Establishes a connection to your local MongoDB server.

```python
client = pymongo.MongoClient("mongodb://localhost:27017/")
db = client["HousingDB"]
collection = db["US_Housing_Market"]
df = pd.read_csv('USA_Housing.csv')
print(list(df))
```

### Step 3: High Income Query
Queries the database for records where the neighborhood's average income is more than $80,000

```python
query_income = collection.find({"Avg_Area_Income": {"$gt": 80000}}, {"_id": 0})
df_high_income = pd.DataFrame(list(query_income))
print({len(df_high_income)})
df_high_income.head()
```

### Step 4: Bedroom Query
Queries the database for real estate records where properties have exactly 3 bedrooms.

```python
query2 = collection.find({"Avg_Area_Number_of_Bedrooms": 3.0}, {"_id": 0})
df_3_beds = pd.DataFrame(list(query2))
print({len(df_3_beds)})
df_3_beds.head()
```

### MongoDB Query - Income And Bedroom Query
Queries for houses with both conditions met
Average area income is greater than $70,000
Average number of bedrooms is 4 or more

```python
query3 = collection.find({
    "Avg_Area_Income": {"$gt": 70000},
    "Avg_Area_Number_of_Bedrooms": {"$gte": 4.0}}, {"_id": 0})

df_combined = pd.DataFrame(list(query3))
print({len(df_combined)})
df_combined.head()
```
